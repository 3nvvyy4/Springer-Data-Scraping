```python
#The scarping here is only for the computer science articles
import requests
res = requests.get("https://link.springer.com/journals/browse-subject?query=&subject=COMPUTER_SCIENCE&sortBy=&page=1")
print(res.status_code)
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(res.text, "html.parser")
```

```python
pagination = soup.find_all("a", class_="eds-c-pagination__link")
numbers = [number.get_text(strip=True) for number in pagination]
print(numbers)
```

```python
page_numbers = []
for elements in range(3):
    page_numbers.append(int(numbers[elements]))
print(page_numbers)
```

```python
numbers_2 = [3,4,5,6,7]
page_numbers = page_numbers + numbers_2
print(page_numbers)
```

```python
page_numbers.sort()
print(page_numbers)
```

```python
all_titles = []
for pages in page_numbers:
    response = requests.get(f"https://link.springer.com/journals/browse-subject?query=&subject=COMPUTER_SCIENCE&sortBy=&page={pages}")
    soup_2 = BeautifulSoup(response.text, "html.parser")
    
    
    
    headers = soup_2.find_all("h2", class_="app-card-open__heading")
    titles = [title.get_text(strip=True) for h in headers for title in h.find_all("span")]

    all_titles.extend(titles)

print(all_titles)
print(len(all_titles))
    
```

```python
import pandas as pd
df = pd.DataFrame(all_titles, columns=["Titles"])
df.to_csv("Articles.csv", index=False, encoding="utf-8-sig")
```
